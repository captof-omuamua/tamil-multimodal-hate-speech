```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_name = "abhinand/tamil-llama-7b-instruct-v0.1"
tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    load_in_8bit=True,  # Use 8-bit quantization; substitute with 4-bit if available
    device_map="auto",
    offload_folder="/home/raptor/Desktop/MODEL_WEIGHTS",  # Folder for CPU offloading
    llm_int8_enable_fp32_cpu_offload=True  # Enable CPU offload for FP32 layers
)

model.eval()
```
